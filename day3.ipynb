```python
class Bank:
    bname='PNB'
    loc='contai'
    ifsc='PNB00002'
    manager_name='gandu'
    def __init__(self,name,age ,phone ,pan,balance):
        self.name=name
        self.age=age
        self.phone=phone
        self.pan=pan
        self.balance=balance
    def display(self):
        print(self.name,self.phone,self.age,self.balance,self.ifsc)
    def change_number(self,change):
        self.phone=change
    def with_money(self,bal):
        if(bal>self.balance):
            print("taka nai")
        else:
            self.balance-=bal
            print("withdraw money",bal)
    def depo_money(self,mon):
        self.balance-=mon
        print("withdraw money",mon)
    @classmethod
    def bank_details(self):
        print(self.bname," :",self.ifsc)
    @classmethod
    def change_ifsc(self,new_ifsc):
        self.ifsc=new_ifsc

u1=Bank('sampad',22,57375634,5743489,1)
u2=Bank('rohan',21,53425345634,57443539,100000000)
```
