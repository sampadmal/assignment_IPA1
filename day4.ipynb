```python
def merge_sor_arr(arr1,arr2):
    i=0
    j=0
    merged=[]
    while i<len(arr1) and j<len(arr2):
        if arr1[i] < arr2[j]:
            merged.append(arr1[i])
            i+=1
        else:
            merged.append(arr2[j])
            j+=1
    while(i<len(arr1)):
        merged.append(arr1[i])
        i+=1
    while(j<len(arr2)):
        merged.append(arr2[j])
        j+=1
    return merged
arr1=[1,2,3,4]
arr2=[4,5,6,7]
print(merge_sor_arr(arr1,arr2))


        
```

```python
def remove_dup(arr):
  fir=[arr[0]]
  for i in range(1,len(arr)):
    if arr[i]!=arr[i-1]:
      fir.append(arr[i])
  return fir

arr=[1,2,2,3,4,4,5]
print(remove_dup(arr))

```

```python
def missing_num(arr,n):
    expected_sum=(n*(n+1))//2
    actual_sum=sum(arr)
    return expected_sum-actual_sum
        
arr=[1,2,3,5,6]  
n=6
print(missing_num(arr,n))   
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Student(Person):
    def __init__(self, name, age,roll,adds):
        super().__init__(name, age)
        self.roll=roll
        self.adds=adds
class teacher(Person):
    def __init__(self, name, age,sub,tid):
        super().__init__(name, age)
        self.sub=sub
        self.tid=tid
class Academic:
    def __init__(self,subject,marks):
        self.subject=subject
        self.marks=marks
class Sports:
    def __init__(self,sport_name,level):
        self.sport_name=sport_name
        self.level=level
class AllRounderStudent(Student,Academic,Sports):
    def __init__(self, name,age,roll,adds,subject, marks,sport_name,level,is_aca,is_spo):
        super().__init__(name,age,roll,adds)
        self.is_aca=is_aca
        self.is_spo=is_spo
        if self.is_aca:
            Academic.__init__(self,subject,marks)
        if self.is_spo:
            Sports.__init__(self,sport_name,level)
    def display(self):
        print(self.name,self.age,self.roll,self.adds)
        if self.is_aca:
            print(self.subject,self.marks)
        if self.is_spo:
            print(self.sport_name,self.level)
std=AllRounderStudent('sampad',22,144,'contai','math',100,'sprint',3,True,False)
std.display()
        
        
```
