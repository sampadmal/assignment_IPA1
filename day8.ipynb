```python
def div(a,b):
    try:
        print("division result",a//b)
    except ZeroDivisionError:
        print("division by zero is not possible")
        b1=int(input("enter second value again"))
        div(a,b1)
    except ValueError:
        print("please enter integer values")
div(10,0)
div(4,2)

```
