```python
list1=[1,6,9,3,7,2]
mid=len(list1)//2
firstSum=sum(list1[ : mid])
secondSum=sum(list1[mid : ])
if(firstSum > secondSum):
    print("First half greater")
elif(firstSum == secondSum):
    print("Equal")
else:
    print("Second half greater")
```
