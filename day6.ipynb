```python
from abc import ABC,abstractmethod
class BankAccount(ABC):
    def __init__(self,cus_name,dob):
        self.cus_name=cus_name
        self.dob=dob
        self.balance=0
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
class SavingsAccount(BankAccount):
    def __init__(self,cus_name,dob,min_bal):
        super().__init__(cus_name,dob)
        self.min_bal=min_bal
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance:
            print("insufficient balance")
        else:
            self.balance-=amount
    def bal_account(self):
        print("current balance:",self.balance)

class CurrentAccount(BankAccount):
    od_limit=5000
    def __init__(self,cus_name,dob,min_bal):
        super().__init__(cus_name,dob)
        #self.balance=self.balance+self.od_limit
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance+self.od_limit:
            print("insufficient balance")
        else:
            self.balance-=amount 
    def bal_account(self):
        print("current balance:",self.balance)

```

```python
c=CurrentAccount('sampad',373,1000)
c.deposit(5000)
c.bal_account()
c.withdraw(11000)
c.bal_account()
```
