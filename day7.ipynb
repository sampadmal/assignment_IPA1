```python
# count number of words in a file
with open('a.txt','r') as f:
    words=f.read().split()
print(len(words))
```

```python
# separate words on the basis of comma 
with open('a.txt','r') as f:
    words=f.read().split(',')
print(len(words))
```

```python
# count all the unique words in a file
with open('a.txt','r') as f:
    words=f.read().split()
    unique_words=set(words)
print(unique_words)
```

```python

with open('a.txt','r') as f:
    words=f.read().split()
    word_count={}
    for i in words:
        if i not in word_count:
            word_count[i]=1
        else:
            word_count[i]+=1
print(word_count)
```

```python
import csv
with open('data.csv','w',newline="") as file:
    filedmanes=['name','age','course']
    writer=csv.DictWriter(file,fieldnames=filedmanes)
    writer.writeheader()
    writer.writerow({'name':'sampad','age':22,'course':'cse'})
    writer.writerow({'name':'rohan','age':20,'course':'cse'})
```

```python
x=np.arange(1,11)
y=np.sin(x)
z=np.cos(x)
print(x,y,z,sep='\n')
```

```python
plt.plot(x,y,label='Sin Theta')
plt.plot(x,z,label='Cos Theta',color='red')
plt.title("Sin Theta And Cos Theta")
plt.xlabel("data points")
plt.ylabel("functions points")
plt.legend()
plt.show()
```
