```python
class A:
    def __init__(self,a):
        self.a=a
    def __add__(self, other):
        return self.a+other.a
    def __sub__(self, other):
        return self.a-other.a
obj1=A(10)
obj2=A(20)
print(obj1+obj2)#internally calls def __add__ function
print(obj1-obj2)#internally calls def __sub__ function
```

```python
class Vehicle:
    def __init__(self,veh_id,brand):
        self.veh_id=veh_id
        self.brand=brand
    def calculate_rent(self,r):
        if r=='Car':
            print("rent on basis of days")
        elif r=='Bike':
            print("rent on basis of hour")
        else:
            print("rent on basis of km")
        
class Car(Vehicle):
    def __init__(self, veh_id, brand,ppd):
        super().__init__(veh_id, brand)
        self.ppd=ppd
    def calculate_rent(self, r):
        print("Rent:",r*self.ppd)
class Bike(Vehicle):
    def __init__(self, veh_id, brand,pph):
        super().__init__(veh_id, brand)
        self.pph=pph
    def calculate_rent(self, r):
        print("Rent:",r*self.pph)
class Truck(Vehicle):
    def __init__(self, veh_id, brand,ppk):
        super().__init__(veh_id, brand)
        self.ppk=ppk
    def calculate_rent(self, r):
        print("Rent:",r*self.ppk)
car=Truck(1,'suzuki',6000)
car.calculate_rent(9)
```
