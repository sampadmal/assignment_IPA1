18) Which country shows the highest growth in total production over the years?

```python
growth = df.sort_values("Year").groupby("Country")["Total Production"] .agg(lambda x: x.iloc[-1] - x.iloc[0])

growth.sort_values(ascending=False).head(1)
```

China has the highest growth in total production over the years

19) Is there any relationship between export price and total production?

```python
df["Export Price"].corr(df["Total Production"])
```

The corelation point between Export Price and Total Production is 0.344 .It shows there is a positive relation between them but the relation is not that strong
