```python
num=1459
even=0
odd=0
while(num!=0):
    r=num%10
    if(r%2==0):
        even=even+1
    else:
        odd=odd+1
    num=num//10
print("even number are ", even )
print("odd number are " ,odd)
```

```python
num=int(input("enter a number"))
rev=0
n=num
while(n!=0):
    r=n%10
    rev=(rev*10)+r
    n=n//10
if(rev==num):
    print("palindrome number")
else:
    print("not a palindrome number")
```
